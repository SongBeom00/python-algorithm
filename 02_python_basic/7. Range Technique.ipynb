#### 7. Range Technique

```python
# 30 ~ -10 까지 -2씩 감소한 결과를 리스트로 출력

result1 = [i for i in range(30, -11, -2)]
print(result1)

result2 = [i for i in range(-10, 31, 2)]
print(list(reversed(result2)))


result3 = [i for i in range(30, -11, -1) if i % 2 == 0]
print(result3)


result4 = [i for i in range(-11, 31) if i % 2 == 0]
print(list(reversed(result4)))

result5 = list(range(30, -11, -2))
print(result5)
```
