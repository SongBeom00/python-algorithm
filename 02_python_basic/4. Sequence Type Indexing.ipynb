#### 4. Sequence Type Indexing

- 시퀀스 자료형(데이터 값이 연속적으로 이루어진 자료구조) : List, Tuple, Str, Range
- List 관련 함수는 중요
- cmp, len, max, min, list, append, extend, index, insert, pop, remove, reverse, sort
- index 함수 대소문자 구분
- sort, sorted 함수 차이 
- sort는 리스트 객체의 메서드, sorted는 내장 함수

- sort() : 리스트 자체를 정렬, 반환값은 None
- sorted() : 정렬된 새로운 리스트 반환, 원본 리스트는 변경되지 않음

```python
x = ['Orange', 'Cherry', 'Apple', 'Kiwi', 'Banana', 'Strawberry']
 
print(f'출력 결과 : {x[4]}')
print('Banana' in x) # True or False 반환

print(x.index('Banana'))
print(x[x.index('Banana')])

print(x[x.index('Banana', 1, len(x))]) # 시작 인덱스와 끝 인덱스를 지정할 수 있다
```
