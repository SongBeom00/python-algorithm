### 2. Assigning & Comparison

- is, not is -> 참조, 객체(same object, 오브젝트) 비교
- ==, != -> 값(value) 비교
- 값, 참조가 같은 비교는 is 사용(중요)

```python
x = 15
y = 25

print(f'x == y : {x == y}')
print(f'x is y : {x is y}')
```

```python
print(id(x))
print(id(y))
```

```python
x = 15
y = 15

print(f'x == y : {x == y}')
print(f'x is y : {x is y}')
```

```python
x = ['orange', 'banana', 'apple'] 
y = x # 파이썬의 리스트(List)는 객체(Object)이기 때문에 같은 주소값을 참조한다.

print(f'x == y {x == y}')
print(f'x is y : {x is y}')
print(id(x))
print(id(y))
```

```python
x = ['orange', 'banana', 'apple']
y = ['orange', 'banana', 'apple']

print(f'x == y {x == y}') # 값은 같지만
print(f'x is y : {x is y}') # 객체(주소값)가 다르다
```

```python
c = []
d = c # 같은 리스트를 참조
e = c + d # 새 리스트 생성
print(c == d)
print(c is d)
print(c == e)
print(c is e)

print(e)
print(id(c))
print(id(e))
```
