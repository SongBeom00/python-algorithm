#### 6. Sequence Item Selection

```python
x = ['grapes', 'mango', 'orange', 'peach', 'apple', 'lime', 
    'banana', 'cherry', 'tomato', 'kiwi', 'blueberry', 'watermelon']

# 리스트에서 apple과 kiwi를 추출해서 대문자로 출력하세요

result1 = []
result1.append(x[x.index('apple')].upper())
result1.append(x[x.index('kiwi')].upper())
print(result1)

result2 = []
for i in x:
    if i == 'kiwi' or i == 'apple':
        result2.append(i.upper())

print(result2)


result3 = list(map(lambda b: b.upper(), filter(lambda a: a == 'apple' or a == 'kiwi', x)))
print(result3)


result4 = [i.upper() for i in x if i == 'apple' or i == 'kiwi']
print(result4)
```
