#### Function Arguments
- Python 에서는 기본값이 있는 매개변수는 기본값이 없는 매개변수보다 뒤에 와야 합니다.
- 모든 인자가 디폴트 값
- 모든 인자가 디폴트 값 X
- 디폴트 값 인자가 뒤로

- * *args → 위치 인자들을 튜플(tuple)로 받음
- * **kwargs → 키워드 인자들을 딕셔너리(dict)로 받음

```python
# def greet(msg='Good morning', name): # 에러 발생
    # return "Hi! " + name + ', ' + msg


def greet(name: str, msg: str='Good morning') -> str:
    return "Hi! " + name + ', ' + msg
print(greet("Kim"))
print(greet("Park", "How do you do?"))


# 가변 인자로 사용
def add(*args):
    total = 0
    for i in args:
        total += i
    return total

print(add(10, 20, 30))

# print(add((i for i in range(1, 101)))) # 제너레이터 1개를 전달해서 에러 발생

print(add(*(i for i in range(1, 101)))) 

def info(**kargs):
    print(kargs)
    
info(name = 'Kim', age = 30)
# info({'Kim' : 30})
data = {'Kim' : 30}
info(**data)
info(**{'Kim' : 30})
```
