#### Dict Items Sum

- Dict : Key-Value 쌍으로 이루어진 자료형
- 자주 사용하는 함수 : get(), items(), keys(), values()


```python
d = {'a' : 17, 'b' : 114, 'c' : 247, 'd' : 362, 'e' : 220, 'f' : 728, 'g' : -283, 'h' : 922}


print(dir(d.values()))

total = 0
for i in d.values():
    total+=i
print(total)


print(sum(d.values()))

print(sum([d[item] for item in d]))

for item in d:
    print(f'key: {item} value {d[item]}')

```

#### Dict Items Add

```python
d = {'a' : 'apple', 'b' : 'grape'}

d.update({'c' : 'banana', 'd' : 'kiwi'})

print(d)
```

```python
d = {'a' : 'apple', 'b' : 'grape'}

d['c'] = 'banana'
d['d'] = 'kiwi'

print(d)
```

#### Dict Data Filtering

```python
# value 값이 25 이상 필터링 후 출력
d = {'a' : 8, 'b' : 33, 'c' : 15, 'd' : 26, 'e' : 12, 'f' : 120}

# 0은 key, 1은 value
result = dict(filter(lambda x:x[1] > 25, d.items()))
print(result)
```

```python
d = {'a' : 8, 'b' : 33, 'c' : 15, 'd' : 26, 'e' : 12, 'f' : 120}

result = {k: v for k, v in d.items() if v > 25}
print(result)
```

```python
d = {'a' : 8, 'b' : 33, 'c' : 15, 'd' : 26, 'e' : 12, 'f' : 120}

result = {}
for k, v in d.items():
    if v > 25:
        result[k] = v
print(result)
```

```python
d = {'a' : 8, 'b' : 33, 'c' : 15, 'd' : 26, 'e' : 12, 'f' : 120}

print(dict({(k, v)for k, v in d.items() if v >  25}))
```

#### Iterate Dict


```python
d = dict(one = list(range(1, 11)), two = list(range(11, 23)), three = list(range(23, 37)))

for k, v in d.items():
    print(f'key "{k}" has values {v} -> total {len(v)}')

def print_dict_summary(d: dict):
    for i in d.keys():
        values = d.get(i)
        print(f"key '{i}' has values {values} -> total : {len(values)}" )
        
print_dict_summary(d)
```
