```python
def sigma_n1(n: int) -> int:
    return (n * ( n + 1 )) // 2

print(sigma_n1(10))

def sigma_n2(n: int) -> int:
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

print(sigma_n2(10))

def sigma_n3(n: int) -> int:
    return sum(range(1, n + 1))

print(sigma_n3(10))
```
