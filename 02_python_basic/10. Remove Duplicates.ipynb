#### Remove Duplicates

- set(집합 자료형) : 중복 허용 X, 순서 X
- list(리스트 자료형) : 중복 허용 O, 순서 O

- OrderedDict : Dict 순서 보장 받을 수 있음 (파이썬 3.7 이상에서는 dict도 순서 보장)

```python
# 리스트에서 중복되는 원소를 제거 후 출력

x = ['a', 1, 'b', 2, 'a', 3, 'b', 4, 5, 'b']

result1 = set(x)
print(list(result1))

# 순서 유지
from collections import OrderedDict

result2 = list(OrderedDict.fromkeys(x))
print(result2)

result3 = []
for i in x:
    if i not in result3:
        result3.append(i)
print(result3) 

result4 = list(dict.fromkeys(x))
print(result4)
```
