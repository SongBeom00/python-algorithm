#### Local Variables

```python
a = 20
def test():
    # 지역 변수
    a = 35
    return a

print('step1 : ', a) # 20

a = 100

print('step2 : ', a) # 100

print('step3 : ', test()) # 35
```

```python
a = 20

def test():
    global a
    
    print(f'result3 결과 : {a}') # 100
    
    a = 35
    
    return a
    
print(f'result1 결과 : {a}') # 20

a = 100

print(f'result2 결과 : {a}') # 100


print(f'result4 결과 : {test()}') # 35


print(f'result5 결과 : {a}') # 35
```
