```python
file_path = './source/22-1.txt'

def cnt_word1(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        txt = file.read()
    txt = txt.replace(',', ' ')
    
    # 쉼표 제거 확인
    print(txt)
    
    txt_list = txt.split()
    print(txt_list)
    return len(txt_list)
    
# in_str = cnt_word1(file_path)
# print(in_str)


import re

def cnt_word2(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        txt = file.read()
    
    # 정규표현식
    txt_list = re.split(' |,', txt)
    
    # 리스트 변환 (공백 + 콤마)
    print(txt_list)
    
cnt_word2(file_path)
```
