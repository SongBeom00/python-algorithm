#### 8. More Range Technique


```python

# 1부터 20까지 홀수 * 10, 짝수는 그대로 출력

result1 = []
for i in range(1, 21):
    if i % 2 != 0:
        result1.append(i * 10)
    else:
        result1.append(i)
print(result1)

result2 = [i * 10 if i % 2 != 0 else i for i in range(1, 21)]
print(result2)


result3 = [[j for j in range(5)] for i in range(5)]
print(result3)
```
